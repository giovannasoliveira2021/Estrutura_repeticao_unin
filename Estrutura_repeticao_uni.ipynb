# Acumuladores

```python
soma = 0
cont = 1
while (cont <= 5):
   x = float(input(f"Digite a {cont} nota: "))
   soma = soma + x
   cont = cont + 1
   media = soma / 5
   print (f"Média final: {media}")
```

Operadores especiais de atribuição

```python
soma = 0
cont = 1
while (cont <= 5):
   x = int(input(f"Digite a {cont} nota: "))
   soma += x # Equivalente: soma = soma + x
   cont += 1 # Equivalente: cont = cont + 1
   print (f"Somatório: {soma}")
```

Validando dados de entrada com um loop

```python
# Validando a entrada
x = int(input("Digite um valor maior do que zero: "))
while(x <= 0):
  x = int(input("Digite um valor maior do que zero: "))
print(f"Você digitou {x}. Encerrando o programa...")
```

Interrompendo um loop com break

```python
print("Digite uma mensagem que irei repetir para você!!")
print("Para encerrar escreva 'sair'.")
while True:
  texto = input("")
  print(texto)
  if texto == "sair":
    break
  print("Encerrando o programa...")
```

Voltando ao inicio do laço com continue

```python
while True:
  nome = input('Qual o seu nome?')
  if (nome != 'Gigi'):
    continue # volta para inicio do laço
  senha = input('Qual a sua senha?')
  if (senha == 'UnInTeR'):
    break # encerra o laço
print('Acesso concedido.')
```

Valores "Truthy" e "Falsey"

```python
nome = "" # Falsey
while not nome: # Not Falsey (Truthy/True)
  # encerra o laço quando nome não estiver vazio
  nome = input('Digite seu nome:')

  valor = int(input('Digite um número qualquer: '))
  if valor: #Equivalente if valor != 0:
    print('Você digitou um valor diferente de zero.')
  else:
    print('Você digitou zero.')
```

Estruturas de repetição aninhadas

```python
# 2x While
tabuada = 1
while tabuada <= 10:
    print(f'TABUADA DO {tabuada}:')
    i = 1
    while i <= 10:
        print(f'{tabuada} x {i} = {tabuada * i}')
        i += 1

    print()
    tabuada += 1
```

```python
# 2x FOR
for tabuada in range (1, 11, 1):
  print(f"TABUADA DO {tabuada}: ")
  for i in range(1, 11, 1):
    print(f"{tabuada} x {i} = {tabuada * i}")
```

```python
# while + for
tabuada = 1
while tabuada <= 10:
  print(f"TABUADA DO {tabuada}:")
  for i in range(1, 11, 1):
    print(f"{tabuada} x {i} = {tabuada * i}")
  tabuada += 1
```
